### Dataset Decomposition & Analysis Plan

**Scope:** February 2019 (Cross-month seasonality and annual trend analysis are out of scope).

| Category | Columns | Planned Engineering & Analytical Goals |
| :--- | :--- | :--- |
| **Temporal** | `duration_sec`<br>`start_time`<br>`end_time` | • Convert duration to minutes; apply log scaling to handle extreme skews.<br>• Bin start minutes (`0–10`, `10–20`, etc.) to evaluate transit connection patterns within the hour.<br>• Cross-tabulate duration against demographics, user type, and equity program enrollment. |
| **Network & Stations** | `start_station_id`<br>`start_station_name`<br>`end_station_id`<br>`end_station_name` | • Identify top origin, destination, and paired travel corridors.<br>• Calculate net station flow ($\text{Arrivals} - \text{Departures}$) to pinpoint bike surplus and deficit hubs.<br>• Flag circular round-trips (`start_station_id == end_station_id`) vs. one-way transit. |
| **Spatial** | `start_station_latitude`<br>`start_station_longitude`<br>`end_station_latitude`<br>`end_station_longitude` | • Calculate straight-line displacement via the Haversine formula.<br>• Approximate speed ($\text{Distance} / \text{Duration}$) to detect non-movement outliers or leisure cruising.<br>• Map geographic trip clusters across San Francisco, Oakland, and San Jose. |
| **Demographics** | `user_type`<br>`member_birth_year`<br>`member_gender` | • Derive user age (`2019 - member_birth_year`) and group into demographic cohorts (e.g., `<25`, `25–34`, `35–54`, `55+`).<br>• Filter implausible birth years (e.g., ages $>80$).<br>• Compare ride duration across gender identities and user subscription models (`Subscriber` vs. `Customer`). |

### Data Hygiene & Null Strategy

* **Station Metadata (`start_station_id`, `end_station_id`):** 197 missing records (~0.1%). Exclude from station-level routing analyses or tag as dockless/off-grid trips; keep for aggregate duration metrics.
* **Demographics (`member_birth_year`, `member_gender`):** 8,265 missing records (~4.5%), concentrated in casual `Customer` trips. Segment demographic analyses exclusively to authenticated `Subscriber` data rather than imputing artificial values.



> ## Missing-data strategy

- **Station columns** (`start_station_id/name`, `end_station_id/name`, 197 rows, ~0.1%): drop via `dropna(subset=...)`. Too few to impute; needed for routing analysis.

- **Demographics** (`member_birth_year`, `member_gender`, 8,265 rows, ~4.5%): **do NOT impute** with mean/mode — that would fabricate ages/genders. Keep as NaN and segment demographic analyses to non-null rows.

- **Duplicates** (4 rows): inspect, then drop (see duplicate cell below).


```python
# Drop the 197 rows missing station id/name (needed for origin/destination analysis).
# This is a drop, not a fill: imputing station names would fabricate routes.
cols_to_drop = ["start_station_id", "start_station_name", "end_station_id", "end_station_name"]
df = df.dropna(subset=cols_to_drop)
```

```python
#check the Nulls 
df.isna().sum()
```

```python
##  still got a missing value in the member_gender, so let's explore this column more 
df[ (df['user_type'] == 'Customer') & (df['member_gender'].isna())].count()
```

```python
# Demographic nulls: quantify by user_type instead of filling with fake values.
print("Remaining birth_year nulls:", int(df["member_birth_year"].isna().sum()))
print("Remaining gender nulls:", int(df["member_gender"].isna().sum()))
print(df.groupby("user_type", observed=True)[["member_birth_year", "member_gender"]].apply(lambda g: g.isna().sum()))
# Strategy: keep NaN; filter to non-null only inside age/gender plots (e.g. df.dropna(subset=['member_age']))
```

```python
# Sanity check: Customers missing gender vs Subscribers missing gender
print("Customers missing gender:", int(((df['user_type'] == 'Customer') & (df['member_gender'].isna())).sum()))
print("Subscribers missing gender:", int(((df['user_type'] == 'Subscriber') & (df['member_gender'].isna())).sum()))
```

```python
## Result: ~3.2k Customers + ~5k Subscribers lack demographics, so missingness is not 'just casual rides'.
print("Total rows after station drop:", len(df))
```
