### Data Hygiene & Null Strategy

* **Station Metadata (`start_station_id`, `end_station_id`):** 197 missing records (~0.1%). Exclude from station-level routing analyses or tag as dockless/off-grid trips; keep for aggregate duration metrics.



> ## Missing-data strategy

- **Station columns** (`start_station_id/name`, `end_station_id/name`, 197 rows, ~0.1%): drop via `dropna(subset=...)`. Too few to impute; needed for routing analysis.

- **Demographics** (`member_birth_year`, `member_gender`, 8,265 rows, ~4.5%): **do NOT impute** with mean/mode — that would fabricate ages/genders. Keep as NaN and segment demographic analyses to non-null rows.

- **Duplicates** (4 rows): inspect, then drop (see duplicate cell below).


```python
# Drop the 197 rows missing station id/name (needed for origin/destination analysis).
# This is a drop, not a fill: imputing station names would fabricate routes.
cols_to_drop = ["start_station_id", "start_station_name", "end_station_id", "end_station_name"]
df = df.dropna(subset=cols_to_drop)
```

```python
# Demographic nulls: quantify by user_type instead of filling with fake values.
print("Remaining birth_year nulls:", int(df["member_birth_year"].isna().sum()))
print("Remaining gender nulls:", int(df["member_gender"].isna().sum()))
print(df.groupby("user_type", observed=True)[["member_birth_year", "member_gender"]].apply(lambda g: g.isna().sum()))
# Strategy: keep NaN; filter to non-null only inside age/gender plots (e.g. df.dropna(subset=['member_age']))
```

 > ## From all previous results above I found  :
 
 
 
 **=========================================================**
 
 

> 1 ) **unique Value** as in ((`member_gender`)) : Not significant , so won't clean it



> 2 ) **Incorrect data type** as in (( `user_type` &&	`member_gender`)) :  significant change , so I need to clean it.

```python
##  step  2 ) Incorrect data type as in (( `user_type` &&	`member_gender`)) :  significant change , so  need to clean it.
df['user_type'] = df['user_type'].astype('category')
df['member_gender'] = df['member_gender'].astype('category')
df['bike_share_for_all_trip'] = df['bike_share_for_all_trip'].astype('category')
```

```python
## check your work
print(df[['user_type','member_gender','bike_share_for_all_trip']].dtypes)
```

```python
## Step 3) Marketing feature: derive rider age from birth year
## Bounds 16-95: 16 = minimum plausible rider age; 95 = generous upper bound.
## Ages >80 flagged as suspicious per analysis plan; >95 treated as data-entry error (e.g. 1878 -> age 141).
df['member_age'] = 2019 - df['member_birth_year']
print(df['member_age'].describe())
# max of 141 is an outlier for sure
```

```python
## Cap implausible ages to NaN (adds ~75 outliers to the ~8263 demographic NaNs)
df['member_age'] = df['member_age'].where((df['member_age'] >= 16) & (df['member_age'] <= 95), np.nan)
print("member_age NaNs (original nulls + ~75 age outliers):", int(df['member_age'].isna().sum()))
```

```python
## Convert age to nullable Int64 (plain int cannot hold NaN)
df['member_age'] = df['member_age'].astype('Int64')
print(df['member_age'].dtype)
print(df['member_age'].describe())
```

```python
## Helper: average / median rider age (skips NaN)
def mean_median_age(s):
    return float(s.mean(skipna=True)), float(s.median(skipna=True))
mean_age, median_age = mean_median_age(df['member_age'].astype('float'))
print(f"Mean age: {mean_age:.1f}, Median age: {median_age:.1f} (non-null n={int(df['member_age'].notna().sum())})")
```

```python
## Trip duration in minutes (more usable than seconds for marketing)
df['duration_min'] = df['duration_sec'] / 60.0
print(df[['duration_sec', 'duration_min']].describe().T)
```

```python
## Trip duration in hours (for long-trip / outlier review)
df['duration_hr'] = df['duration_sec'] / 3600.0
print(df[['duration_sec', 'duration_min', 'duration_hr']].describe().T)
```

```python
## Verify new columns: dtypes, null counts, sanity ranges
print(df[['member_age', 'duration_min', 'duration_hr']].dtypes)
print(df[['member_age', 'duration_min', 'duration_hr']].isna().sum())
print("duration_min > 30 min share:", float((df['duration_min'] > 30).mean()))
```

```python
## Q3: distribution of rider ages (right-skewed; mean > median)
a = df['member_age'].astype('float')
print(a.describe().to_string())
print('Mean / median:', round(float(a.mean()),1), '/', round(float(a.median()),1))
plt.figure(figsize=(8,4))
sns.histplot(a.dropna(), bins=50, kde=True)
plt.axvline(a.median(), color='red', linestyle='--', label='median')
plt.title('Age distribution (non-null riders)')
plt.xlabel('Age'); plt.ylabel('Rides'); plt.legend()
plt.tight_layout(); plt.show()
```


# <a id="4">  4 ) . Do the users' ages have outlier values?</a>


> ### <mark>Your analysis and exploration here </mark>

```python
## Q4: do ages have outliers? boxplot + IQR rule + cap audit
a = df['member_age'].astype('float').dropna()
q1, q3 = a.quantile(0.25), a.quantile(0.75)
iqr = q3 - q1
lo, hi = q1 - 1.5*iqr, q3 + 1.5*iqr
print(f'IQR bounds: [{lo:.1f}, {hi:.1f}]')
print('Beyond IQR:', int(((a < lo) | (a > hi)).sum()), '| Age > 80:', int((a > 80).sum()))
plt.figure(figsize=(7,3))
sns.boxplot(x=a)
plt.title('Age outliers (boxplot, non-null)')
plt.xlabel('Age')
plt.tight_layout(); plt.show()
# NOTE: raw max was 141 (birth year 1878); values >95 already capped to NaN in cleaning.
```

# <a id="1100"> Conclusions </a>

## Key findings (Feb 2019, n=183,215 after cleaning)

- **Subscriber system:** 89.2% Subscriber (163,414) vs 10.8% Customer (19,801). Core 25-34 cohort is ~47-49% of rides in both groups; 55+ is small (Subscribers 4.6%, Customers 3.4%).

- **Gender split (non-null):** Male 130,500 / Female 40,805 / Other 3,647, plus 8,263 missing. Female riders are youngest on average (mean 33.2, median 31) vs Male (34.4 / 32) and Other (35.9 / 34).

- **Age:** mean 34.2, median 32.0, right-skewed. Raw max 141 (birth year 1878) is a data-entry error; 75 values >95 were capped to NaN, leaving 117 riders >80. IQR bounds [9.0, 57.0] flag 5,706 statistical outliers, but most are legitimate older riders — only >95 was treated as error.

- **Trip length:** mean 12.1 min, median 8.6 min; 95th pct 26.2 min, 99th pct 57.6 min; only 3.68% of trips exceed 30 min. Max 1,424 min (23.7 h) indicates overnight/leisure/non-return outliers, not typical use.

- **Who rides longer:** Customers average 23.9 min (median 13.2) vs Subscribers 10.7 min (median 8.2) — casual users take ~2x longer trips. By gender, Other has the highest mean (16.6 min), then Female (13.0), then Male (11.2); medians are closer (9.5 / 9.3 / 8.2).

- **Equity program:** 9.5% of rides enrolled (17,346). Enrollment is 0% among Customers and 10.6% among Subscribers, so the program reaches committed members, not casual riders.

- **Correlations:** effectively no linear link between duration and demographics, bike_id, or station coordinates (|r| < 0.05). Start/end coordinates correlate ~0.99 (short hops); longitude-latitude ~-0.67 reflects Bay Area geography. `birth_year vs age = -1.0` is definitional (`age = 2019 - birth_year`), and `sec/min/hr` are 100% linked by construction — neither is a finding.

## Limitations

- **Single month (Feb 2019):** no seasonality, annual-trend, or weather/holiday analysis possible.

- **Corrupt timestamps in source CSV:** `start_time`/`end_time` contain only `MM:SS.ms` (e.g. `32:10.1`), no date/hour, so time-of-day, rush-hour, and dwell-time analysis were out of scope and left untouched.

- **Missing demographics (~4.5%, 8,263 rows + 75 age caps = 8,338 age NaNs):** present in both Customers (3,235) and Subscribers (5,028), so results by age/gender describe only reporting riders; values were retained as NaN, never imputed.

- **Duration outliers retained:** extreme long trips inflate means — medians and 99th-pct clipping were used for plots, but no trips were deleted for being long.

- **Station coordinates are dock locations, not GPS traces:** true route distance, speed, and circular-trip detection need Haversine displacement on top of duration, not done here.

- **Equity flag only:** `bike_share_for_all_trip` records enrollment, not income or eligibility, so behavioral differences are descriptive, not causal.
